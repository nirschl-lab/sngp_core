```python
acevedo_data = {'file_name': [], 'label': [], 'fold': []}
```

```python
for item in data['train']:
    file_name = item['filename']
    label_name = item['label_name']
    acevedo_data['file_name'].append(file_name)
    acevedo_data['label'].append(label_name)
    acevedo_data['fold'].append('train')
```

```python
for item in data['validation']:
    file_name = item['filename']
    label_name = item['label_name']
    acevedo_data['file_name'].append(file_name)
    acevedo_data['label'].append(label_name)
    acevedo_data['fold'].append('validation')
```

```python
for item in data['test']:
    file_name = item['filename']
    label_name = item['label_name']
    acevedo_data['file_name'].append(file_name)
    acevedo_data['label'].append(label_name)
    acevedo_data['fold'].append('test')
```

```python
acevedo_total_csv = pd.DataFrame(acevedo_data)
```

```python
#subset dataset where lablel=basophil, eosinophil, monocyte
acevedo_wbc_subset = acevedo_total_csv[acevedo_total_csv['label'].isin(['basophil', 'eosinophil', 'monocyte'])]
```

```python
#remane column label to acevedo_label
acevedo_wbc_subset.rename(columns={'label': 'acevedo_label'}, inplace=True)
```

```python
# print(len(acevedo_wbc_subset))
# print(acevedo_wbc_subset['acevedo_label'].value_counts())
# acevedo_wbc_subset.head()
#map to wbc labels
wbc_map = {'basophil': 'BA', 'eosinophil': 'EO', 'monocyte': 'MO'}
acevedo_wbc_subset['labels'] = acevedo_wbc_subset['acevedo_label'].map(wbc_map)
```
