```python
import pandas as pd
```

```python
weight_resampled = pd.read_csv('results/resnet18_weighted_sampling_5epoch.csv')
vanilla  = pd.read_csv('results/resent18_label_smooth.csv')
```

```python
weight_resampled.head()
```

```python
#if lables in ['SNE', 'LY', 'MO'] select from vanilla or select from weight_resampled
voted_csv = {'ID': [], 'labels': []}
for id in weight_resampled['ID']:
    if vanilla[vanilla['ID']==id]['labels'].item() in ['SNE', 'LY', 'MO']:
        voted_csv['ID'].append(id)
        voted_csv['labels'].append(vanilla[vanilla['ID']==id]['labels'].item())
    else:
        voted_csv['ID'].append(id)
        voted_csv['labels'].append(weight_resampled[weight_resampled['ID']==id]['labels'].item())
```

```python
pd.DataFrame(voted_csv).to_csv('results/voted_results.csv', index=False)
```
