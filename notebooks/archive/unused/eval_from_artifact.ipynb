```python
li = [11, 12, 13, 14, 1, 2, 3]
```

```python

pile = [0 for _ in range(len(li))]

curr_size = 0
for num in li:
    i=0; j=curr_size
    while i != j:
        mid = (i+j)//2
        if num <= pile[mid]:
            j = mid
        else:
            i = mid + 1
    pile[i] = num
    curr_size = max(i+1, curr_size)
```
