```python
import torch.nn as nn
import torch
import timm
```

```python
backbone_name = 'vit_small_patch16_224.dino' #'resnet50' #'vit_small_patch16_dinov3_qkvb.lvd_1689m' #'dinov2_vit_S14'
pretrained = True
backbone = timm.create_model(
            backbone_name,
            pretrained=pretrained,
            num_classes=0,       # <- critical
                # works for CNNs; for ViTs you can also try "token" / "avg"
        )

batch_size = 2
in_chans = 3
height = width = 224
x = torch.randn(batch_size, in_chans, height, width)
out = backbone(x)
print(out.shape)
```

```python
backbone_name = 'resnet50' #'vit_small_patch16_224.dino' #'vit_small_patch16_dinov3_qkvb.lvd_1689m' #'dinov2_vit_S14'
pretrained = True
backbone = timm.create_model(
            backbone_name,
            pretrained=pretrained,
            num_classes=0,       # <- critical
            global_pool='avg'   # works for CNNs; for ViTs you can also try "token" / "avg"
        )

batch_size = 2
in_chans = 3
height = width = 224
x = torch.randn(batch_size, in_chans, height, width)
out = backbone(x)
print(out.shape)
```

```python
out = backbone.forward_features(x)
print(out.shape)
```

```python
num_classes=10
in_chans=3
model_name = 'vit_small_patch16_224.dino' #'vit_small_patch16_dinov3_qkvb.lvd_1689m' #'dinov2_vit_S14'
pretrained = True
model = timm.create_model(model_name,
            pretrained=pretrained,
            num_classes=num_classes,
            in_chans=in_chans)
```
