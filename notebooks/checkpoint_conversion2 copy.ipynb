### Acevedo Baseline

```python
acevedo_baseline_classifier = BaselineClassifier(
    arch = "resnet18",
    num_classes = 8,
    dropout_p = 0.5,
    pretrained = False,)

acevedo_baseline_classifier.load_state_dict(torch.load(acevedo_baseline_ckpt_path, map_location=device))
print("Acevedo Baseline Classifier loaded successfully.")

#test input
 # Example input tensor
acevedo_baseline_classifier.eval()  # Set the model to evaluation mode
with torch.no_grad():
    logits = acevedo_baseline_classifier(input_tensor)

print("Logits shape:", logits.shape)
```

### Acevedo SNGP

```python
acevedo_sngp_classifier = SNGPClassifier(num_classes = 8,
  arch = "resnet18",
  pretrained = False,         
  rff_dim = 1024,       
  length_scale = 1.0,
  ridge_penalty = 1e-3,
  cov_momentum = 0.999,
  mean_field = True)

acevedo_sngp_classifier.load_state_dict(torch.load(acevedo_sngp_ckpt_path, map_location=device))
print("Acevedo SNGP Classifier loaded successfully.")


#test input
 # Example input tensor
acevedo_sngp_classifier.eval()  # Set the model to evaluation mode
with torch.no_grad():
    mean_field_logits, raw_logits, pred_var = acevedo_sngp_classifier(input_tensor)

print("Mean field logits shape:", raw_logits.shape)
```
